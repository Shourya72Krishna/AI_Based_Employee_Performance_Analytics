```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

MODELS = {
    "qwen": "Qwen/Qwen2.5-1.5B",
    "llama": "meta-llama/Llama-3.2-3B"
}

for name, path in MODELS.items():
    tok = AutoTokenizer.from_pretrained(path)
    model = AutoModelForCausalLM.from_pretrained(
        path, output_hidden_states=True, torch_dtype=torch.float16
    ).to("cuda")
    inputs = tok("The capital of France is", return_tensors="pt").to("cuda")
    with torch.no_grad():
        out = model(**inputs)
    print(name, "num hidden layers:", len(out.hidden_states), "shape:", out.hidden_states[0].shape)
```
